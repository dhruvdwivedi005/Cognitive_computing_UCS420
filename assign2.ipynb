```python
a=input("Enter your roll no.")
L=[]

for i in a:
    temp=int(i)*10
    L.append(temp)

print(L) #first ques

L.append(1000) #added 1000 at last with append
print(L)

L.insert(0,1000) #inserted 1000 at first (0th) index using insert
print(L) #second ques

L.remove(1000) #remove only first occurance
print(L)

L.pop() #remove last element
print(L) #third ques

L.sort() #sort in increasing order
print(L)

L.sort(reverse=True) #sort in decreasing order
print(L) #fourth ques

print(L[0:3]) #first three elements
print(L[-3:]) #last three elements

avg=sum(L)/len(L)
new=[i for i in L if i>avg]
print(new) #elements greater than average
```

```python
scores=tuple(L[0:8])
print(scores)

highest=max(scores)
print("highest score ",highest)
print("index of highest ",scores.index(highest))

lowest=min(scores)
print("lowest ",lowest)
print("count of lowest ",scores.count(lowest))

reversed_list=list(scores[::-1])
print(reversed_list)
#tuple cannot be changed directly because tuples are immutable

score=int(input("value to check"))

if score in scores:
    print("first occurence ",scores.index(score))
else:
    print("not exist")

try:
    scores[0]=500
except Exception as e:
    print(e)
    #tuple cannot be changed after creation, unlike a list

first,second,*remaining=scores
print(first)
print(second)
print(remaining)
```

```python
import random

roll_no=int(a)

random.seed(roll_no)

numbers=[]

for i in range(100):
    numbers.append(random.randint(100,900))

print(numbers)

odd=[i for i in numbers if i%2!=0]
print("odd numbers ",len(odd))
print(odd)

even=[i for i in numbers if i%2==0]
print("even numbers ",len(even))
print(even)

def prime(n):
    if n<2:
        return False

    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False

    return True

primes=[i for i in numbers if prime(i)]

print("number of prime numbers ",len(primes))
print(primes)

most=numbers[0]
count=0

for i in numbers:
    if numbers.count(i)>count:
        most=i
        count=numbers.count(i)

print("most frequent number ",most)
print("number of times ",count)
```

```python
digits=a[0:8]

A=set()
B=set()

for i in digits:
    A.add(int(i)*7)
    B.add(int(i)*9)

print("A = ",A)
print("B = ",B)

print("union = ",A.union(B))

print("intersection = ",A.intersection(B))

print("A-B = ",A.difference(B))
print("B-A = ",B.difference(A))
#difference gives the values present in one set but not the other,
#while symmetric_difference gives all values that are not common

print("symmetric difference = ",A.symmetric_difference(B))

print("A is subset of B ",A.issubset(B))
print("B is superset of A ",B.issuperset(A))

X=int(input("Enter value to remove "))

A.discard(X)
print(A)
#discard does not give an error if the value is not present
```

```python
my_dict={
    "name":"Dhruv Dwivedi",
    "roll_no":1024170243,
    "branch":"Computer Science",
    "age":20,
    "city":"prayagraj"
}

print(my_dict)

value=my_dict.pop("city")
my_dict["location"]=value
print(my_dict) #city renamed to location

my_dict["cgpa"]=9.67
print(my_dict)

my_dict["age"]=my_dict["age"]+1
print(my_dict)

dict1=my_dict.copy()
dict1.pop("branch")
print(dict1)

dict2=my_dict.copy()
del dict2["branch"]
print(dict2)

#pop returns the deleted value while del only deletes the key

for key,value in my_dict.items():
    print(key,"->",value)

if "email" in my_dict:
    print(my_dict["email"])
else:
    print("email does not exist")

friend_dict={
    "name":"Rahul",
    "roll_no":"1234567890",
    "branch":"Computer Science",
    "age":20,
    "city":"Delhi"
}

new_dict={**my_dict,**friend_dict}
print(new_dict)

#when both dictionaries have the same key,
#the value from the second dictionary wins

string_dict={key:value for key,value in my_dict.items() if type(value)==str}
print(string_dict)
```
