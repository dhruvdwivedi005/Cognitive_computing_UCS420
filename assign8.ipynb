```python
import re
from nltk.tokenize import sent_tokenize
import nltk

original_text = """Cats are one of the most fascinating and adorable animals in the world. They have soft fur, sparkling eyes, and playful personalities that make them wonderful companions. I love how cats enjoy chasing toys, climbing trees, and curling up in warm places for a peaceful nap. Even though they can be independent and mysterious, they show affection by purring, rubbing against their owners, and sitting close to them. Cats also have amazing senses and graceful movements that make them excellent hunters. In my opinion, cats bring happiness, comfort, and love into our lives with their cute and charming behavior."""

#lower case and punc
text = original_text.lower()
text = re.sub(r'[^\w\s]', '', text)

print("Processed text:")
print(text)

#into sentence
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
sentences = sent_tokenize(original_text)
print("\nSentence tokens:")
for sentence in sentences:
    print(sentence)
```

```python
token = text.split()
print("Word tokens using split():", token)
print("Number of tokens using split():", len(token))
```

```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

token_nltk = word_tokenize(text)
print("Using word_tokenize():", token_nltk)
print("Number of tokens using word_tokenize():", len(token_nltk))

print("\nComparison:")
print("split() separates text wherever whitespace occurs.")
print("word_tokenize() uses NLTK tokenization rules and can handle punctuation as separate tokens.")
print("Since punctuation was removed before this comparison, both methods will usually give similar results here.")
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

filtered_words = [word for word in token_nltk if word.lower() not in stop_words]
print("Words after removing stopwords:", filtered_words)

```

```python
import pandas as pd
word_freq = {}

for word in filtered_words:
    if word in word_freq:
        word_freq[word] += 1
    else:
        word_freq[word] = 1

print("Word Frequency Distribution:")
print(word_freq)

df = pd.DataFrame.from_dict(word_freq, orient='index').reset_index()
df.columns = ['Words', 'Frequency']

df
```

```python
import re
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords

nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

# Extract words containing only alphabets using re.findall()
alphabetic_words = re.findall(r'\b[a-zA-Z]+\b', text)
print("Words containing only alphabets:", alphabetic_words)

# Remove stopwords using NLTK's stopword list
stop_words = set(stopwords.words('english'))
filtered_words = [word.lower() for word in alphabetic_words if word.lower() not in stop_words]
print("\nWords after removing stopwords:", filtered_words)

# Stemming
ps = PorterStemmer()
stemmed_words = [ps.stem(word) for word in filtered_words]
print("\nStemmed words:", stemmed_words)

# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]
print("\nLemmatized words:", lemmatized_words)

print("\nComparison:")
print("Stemming cuts words to a root-like form and may produce non-dictionary words.")
print("Lemmatization returns dictionary forms, but the default lemmatizer treats words as nouns unless a part of speech is provided.")
print("For fast search/indexing, stemming can be useful. For readable text analysis where meaningful words matter, lemmatization is usually preferable.")
```

```python
#%pip install scikit-learn

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

texts = [
    "Cats love playing with toys and sleeping peacefully",
    "Python is a popular programming language",
    "Good phones have excellent cameras and batteries"
]

# Bag of Words representation
cv = CountVectorizer()
bow = cv.fit_transform(texts)

print("Bag of Words Representation:")
print("Vocabulary:", cv.get_feature_names_out())
print(bow.toarray())

# TF-IDF representation
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(texts)

print("\nTF-IDF Representation:")
print("Vocabulary:", tfidf.get_feature_names_out())
print(tfidf_matrix.toarray())

features = tfidf.get_feature_names_out()

print("\nTop 3 Keywords from Each Text:")
for i, row in enumerate(tfidf_matrix.toarray()):
    top_indices = row.argsort()[-3:][::-1]
    print(f"Text {i+1}: {texts[i]}")
    keywords = []
    for index in top_indices:
        keywords.append(features[index])
        print(features[index], ":", round(row[index], 3))
    print("Interpretation: These words have the highest TF-IDF scores in this text and help distinguish its main topic.")
```

```python
import re
from sklearn.metrics.pairwise import cosine_similarity

text1 = """Artificial Intelligence enables machines to learn from data.
AI uses algorithms to solve problems and make decisions.
It supports automation, prediction, and intelligent systems.
AI is widely used in healthcare and education."""

text2 = """Blockchain stores data in a secure digital ledger.
It uses cryptography to protect transactions.
Blockchain supports transparency and decentralized systems.
It is widely used in finance and digital currencies."""

def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)
    tokens = text.split()
    return tokens

tokens1 = preprocess(text1)
tokens2 = preprocess(text2)

print("AI Tokens:", tokens1)
print("\nBlockchain Tokens:", tokens2)

set1 = set(tokens1)
set2 = set(tokens2)

jaccard = len(set1 & set2) / len(set1 | set2)
print("\nJaccard Similarity:", round(jaccard, 3))

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform([text1, text2])

cosine = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])
print("Cosine Similarity:", round(cosine[0][0], 3))

print("\nAnalysis of similarity metrics:")
print("Jaccard similarity compares the overlap of unique words and ignores word frequency.")
print("Cosine similarity compares TF-IDF vectors, giving more weight to words that are informative in the texts.")
if cosine[0][0] > jaccard:
    print("In this example, cosine similarity is higher than Jaccard similarity, suggesting TF-IDF weighting captures some shared informative vocabulary beyond simple set overlap.")
elif cosine[0][0] < jaccard:
    print("In this example, Jaccard similarity is higher than cosine similarity; the unique-word overlap is relatively stronger than the weighted TF-IDF alignment.")
else:
    print("Both metrics give the same similarity score in this example.")
print("For these technology descriptions, cosine similarity generally gives richer insight because it accounts for TF-IDF weights; Jaccard is simpler and easier to interpret as set overlap.")
```

```python
#%pip install textblob wordcloud matplotlib

from textblob import TextBlob
import matplotlib.pyplot as plt
from wordcloud import WordCloud

#review
reviews = [
    "The headphones deliver excellent sound quality and the battery lasts all day. I am very happy with this purchase.",
    "The laptop is fast, lightweight, and perfect for programming. The keyboard feels comfortable and the display is beautiful.",
    "This water bottle is amazing. It keeps my drinks cold for hours and is easy to carry everywhere.",
    "The phone battery drains quickly and the camera produces blurry pictures. I am disappointed with the overall quality.",
    "The headphones stopped working after two days. The customer service was unhelpful and the product was overpriced.",
    "The delivery was late and the package arrived damaged. I would not recommend this product to anyone.",
    "The notebook has 200 pages and a blue cover. It is available in several sizes.",
    "The mouse has three buttons and connects through a USB cable. It works with my computer.",
    "The package arrived yesterday. The product includes a charger and a user manual."
]

positive_reviews = []

#polarity, subjectivity and classify reviews

for review in reviews:
    blob = TextBlob(review)

    polarity = blob.sentiment.polarity
    subjectivity = blob.sentiment.subjectivity

    if polarity > 0:
        sentiment = "Positive"
        positive_reviews.append(review)
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print("\nReview:", review)
    print("Polarity:", round(polarity, 2))
    print("Subjectivity:", round(subjectivity, 2))
    print("Sentiment:", sentiment)

#word cloud
text = " ".join(positive_reviews)

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white"
).generate(text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud of Positive Reviews")
plt.show()
print("The word cloud displays the most frequent words in the positive reviews; larger words occur more often.")
```
