```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Original array:", arr)
print("After adding 2:", arr + 2)
print("After multiplying by 3:", arr * 3)
print("After dividing by 2:", arr / 2)
```

```python
arr1 = np.array([1, 2, 3, 6, 4, 5])
print("Original array:", arr1)
print("Reversed array:", arr1[::-1])
```

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])

values, counts = np.unique(x, return_counts=True)

index = np.argmax(counts)
most_frequent = values[index]
indices = np.where(x == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Frequency:", counts[index])
print("Indices:", indices)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
max_count = np.max(counts)
most_frequent = values[counts == max_count]

print("Most frequent values:", most_frequent)
print("Frequency:", max_count)

for value in most_frequent:
    indices = np.where(y == value)[0]
    print("Indices of", value, ":", indices)
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])
```

```python
Dhruv = np.linspace(10, 100, 25)

print("Array:", Dhruv)
print("Dimensions:", Dhruv.ndim)
print("Shape:", Dhruv.shape)
print("Total elements:", Dhruv.size)
print("Data type:", Dhruv.dtype)
print("Total bytes:", Dhruv.nbytes)

Dhruv_transpose = Dhruv.reshape(1, 25).T
print(Dhruv_transpose)
```

```python
ucs420_dhruv = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original array:")
print(ucs420_dhruv)

print("Mean:", np.mean(ucs420_dhruv))
print("Median:", np.median(ucs420_dhruv))

print("Maximum:", np.max(ucs420_dhruv))
print("Minimum:", np.min(ucs420_dhruv))

print("Unique elements:", np.unique(ucs420_dhruv))
```

```python
reshaped_ucs420_dhruv = ucs420_dhruv.reshape(4, 3)

print("Reshaped array:")
print(reshaped_ucs420_dhruv)


resized_ucs420_dhruv = np.resize(ucs420_dhruv, (2, 3))

print("Resized array:")
print(resized_ucs420_dhruv)
```
