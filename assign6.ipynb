```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a. Add 2°C to every reading
temperature_corrected = temperature + 2

print("Original temperature:", temperature)
print("Temperature after adding 2°C:", temperature_corrected)

# b. Convert Celsius to Fahrenheit
fahrenheit = (9 / 5) * temperature_corrected + 32

print("Temperature in Fahrenheit:", fahrenheit)

# c. Identify readings greater than 32°C
readings_greater_32 = temperature_corrected[temperature_corrected > 32]

print("Readings greater than 32°C:", readings_greater_32)

# d. Count readings greater than 32°C
count = np.sum(temperature_corrected > 32)

print("Number of readings greater than 32°C:", count)

# e. Vectorization and Boolean indexing
print("Vectorization performs operations on the complete array without using a for loop.")
print("Boolean indexing selects only the elements that satisfy a given condition.")
```

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a. Total steps
total_steps = np.sum(steps)
print("Total steps:", total_steps)

# b. Mean number of steps
mean_steps = np.mean(steps)
print("Mean steps:", mean_steps)

# c. Maximum and minimum values
maximum = np.max(steps)
minimum = np.min(steps)
print("Maximum steps:", maximum)
print("Minimum steps:", minimum)

# d. Total steps for each day
total_each_day = np.sum(steps, axis=0)
print("Total steps for each day:", total_each_day)

# e. Total steps for each user
total_each_user = np.sum(steps, axis=1)
print("Total steps for each user:", total_each_user)

# f. Position of maximum number of steps
max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("Position of maximum steps:", max_position)
```

```python
# a. Create original array
original = np.array([1, 2, 3, 4, 5, 6])

# b. Create slice from index 1 to 4
subset = original[1:5]
print("Original array:", original)
print("Subset:", subset)

# c. Modify first element of subset
subset[0] = 999
print("After modifying subset:")
print("Original:", original)
print("Subset:", subset)

# d. Create another slice using copy()
copied_array = original[1:5].copy()
copied_array[0] = 500
print("After modifying copied array:")
print("Original:", original)
print("Copied array:", copied_array)

# e. Create array from 1 to 12 and reshape into 3x4
matrix = np.arange(1, 13).reshape(3, 4)
print("3 x 4 matrix:")
print(matrix)

# f. Indexing and slicing

# First row
print("First row:", matrix[0, :])

# Last row
print("Last row:", matrix[-1, :])

# Second column
print("Second column:", matrix[:, 1])

# Rows 1-2 and columns 2-3
print("Rows 1-2 and columns 2-3:")
print(matrix[0:2, 1:3])

# g. Flatten using flatten()
flattened = matrix.flatten()
print("Flattened array using flatten():", flattened)

# Flatten using ravel()
raveled = matrix.ravel()
print("Flattened array using ravel():", raveled)

# h. Modify ravel() array
raveled[0] = 100
print("After modifying ravel():")
print("Ravel array:", raveled)
print("Original matrix:")
print(matrix)

# i. Modify flatten() array
flattened[1] = 200
print("After modifying flatten():")
print("Flatten array:", flattened)
print("Original matrix:")
print(matrix)

# j. Display shape, ndim, size and dtype
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# a. Shape and dimensions
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

# b. Transpose
X_transpose = X.T
print("X.T:")
print(X_transpose)

# c. Matrix product X.T @ X
XtX = X.T @ X
print("X.T @ X:")
print(XtX)

# d. Matrix inverse
XtX_inverse = np.linalg.inv(XtX)
print("Inverse of X.T @ X:")
print(XtX_inverse)

# e. Ordinary Least Squares
beta = XtX_inverse @ X.T @ y
print("Beta coefficients:")
print(beta)

# f. Coefficients
print("Sleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])

# g. Prediction for new user
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("Predicted assistance score:", predicted_score)
```
