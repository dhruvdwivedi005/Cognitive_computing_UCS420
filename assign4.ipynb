```python
import pandas as pd

fixed_entries = [
    {"question": "what is the annual fee", 
     "answer": "The annual fee is Rs 500.",
     "keywords": "fee cost price charge", 
     "category": "billing"},

    {"question": "how to reset password", 
     "answer": "Go to Settings > Reset Password.",
     "keywords": "password reset login", 
     "category": "account"},

    {"question": "what are your working hours", 
     "answer": "We are open 9 AM to 5 PM.",
     "keywords": "hours timing open time", 
     "category": "general"},

    {"question": "how can i pay the fee", 
     "answer": "You can pay via UPI, card, or net banking.",
     "keywords": "pay payment upi fee", 
     "category": "billing"},
]

personalized_entries = [
    {"question": "how do i update my registered mobile number",
     "answer": "Go to Account Settings and update your registered mobile number.",
     "keywords": "mobile number update account",
     "category": "account"},

    {"question": "how can i check my refund status",
     "answer": "You can check your refund status from the billing section.",
     "keywords": "refund status billing payment",
     "category": "billing"}
]

all_entries = fixed_entries + personalized_entries

df = pd.DataFrame(all_entries)

print(df)
```

```python
def score_query(query, df):
    query_words = query.lower().split()

    results = []

    for index, row in df.iterrows():
        keywords = row["keywords"].lower().split()

        score = 0

        for word in query_words:
            if word in keywords:
                score += 1

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    results = sorted(results, key=lambda x: x["score"], reverse=True)

    return results
```

```python
query = input("Enter your question: ")

results = score_query(query, df)

for result in results:
    print(result)
```

```python
def same_category(category_name, df):
    return df[df["category"] == category_name]

result = same_category("account", df)

print(result)
```

```python
def score_query_tie(query, df):
    query_words = query.lower().split()

    results = []

    for index, row in df.iterrows():
        keywords = row["keywords"].lower().split()

        score = 0

        for word in query_words:
            if word in keywords:
                score += 1

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    results = sorted(results, key=lambda x: x["score"], reverse=True)

    if len(results) == 0:
        print("No matching entry found.")
        return

    highest_score = results[0]["score"]

    top_results = []

    for result in results:
        if result["score"] == highest_score:
            top_results.append(result)

    print("Best matching entries:")

    for result in top_results:
        print(result)
```

```python
print("Tie Query:")
score_query_tie("fee", df)
```

```python
print("Non-Tie Query:")
score_query_tie("password", df)
```
